````python
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline

import joblib

RANDOM_STATE = 42

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
````

````python
# Read the uploaded csv file into a DataFrame
df = pd.read_csv('creditcard.csv')

# Display the DataFrame info
df.info()
````

````python
# Call describe() as a function to see statistical summary
df.describe()
````

````python
# Check for missing values in each column
print("Missing Values:")
print(df.isna().sum())

# Check for duplicate rows
print("\nDuplicate Rows:")
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")
````

````python
# Drop duplicate rows in place
df.drop_duplicates(inplace=True)

# Confirm that duplicates have been removed
print(f"Remaining rows after dropping duplicates: {len(df)}")
print(f"Remaining duplicate rows: {df.duplicated().sum()}")
````

# ## Exploratory Data Analysis & Visualizations
Before splitting our data and modeling, we perform several exploratory visualizations to study target imbalance, feature boundaries, class densities, and high-dimensional projections.

````python
# 1. Visualize the Target Variable 'Class' Distribution
plt.figure(figsize=(6, 4))
sns.countplot(x='Class', data=df, palette='Set1', hue='Class', legend=False)
plt.title('Distribution of Transaction Classes (0: Genuine, 1: Fraudulent)')
plt.xlabel('Class')
plt.ylabel('Count (Log Scale)')
plt.yscale('log')
plt.show()
````

````python
# 2. Class Distribution Density for Scaled Time and Amount Columns
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.kdeplot(data=df[df['Class'] == 0], x='Scaled_Amount', label='Genuine (0)', fill=True, ax=axes[0], color='#1f77b4', common_norm=False)
sns.kdeplot(data=df[df['Class'] == 1], x='Scaled_Amount', label='Fraud (1)', fill=True, ax=axes[0], color='#d62728', common_norm=False)
axes[0].set_title('Distribution of Scaled Transaction Amount by Class')
axes[0].set_xlabel('Scaled Amount')
axes[0].set_ylabel('Density')
axes[0].set_xlim(-2, 10)  # Zoom in slightly to view the core distribution
axes[0].legend()

sns.kdeplot(data=df[df['Class'] == 0], x='Scaled_Time', label='Genuine (0)', fill=True, ax=axes[1], color='#1f77b4', common_norm=False)
sns.kdeplot(data=df[df['Class'] == 1], x='Scaled_Time', label='Fraud (1)', fill=True, ax=axes[1], color='#d62728', common_norm=False)
axes[1].set_title('Distribution of Scaled Transaction Time by Class')
axes[1].set_xlabel('Scaled Time')
axes[1].set_ylabel('Density')
axes[1].legend()

plt.tight_layout()
plt.show()
````

````python
# 3. 2D Scatter Plot of Important Features (V17 vs V14)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='V17',
    y='V14',
    hue='Class',
    data=df.sample(50000, random_state=RANDOM_STATE),
    palette={0: '#1f77b4', 1: '#d62728'},
    alpha=0.6
)
plt.title('Transaction Separation across Two High-Impact Features (V17 vs V14)')
plt.xlabel('V17')
plt.ylabel('V14')
plt.legend(title='Class', labels=['Genuine (0)', 'Fraud (1)'])
plt.show()
````

````python
# 4. t-SNE 2D Projection of Transactions (Subsampled for Speed)
from sklearn.manifold import TSNE

fraud_cases = df[df['Class'] == 1]
genuine_cases = df[df['Class'] == 0].sample(n=len(fraud_cases) * 3, random_state=RANDOM_STATE)

tsne_data = pd.concat([fraud_cases, genuine_cases])
X_tsne = tsne_data.drop('Class', axis=1)
y_tsne = tsne_data['Class']

tsne = TSNE(n_components=2, random_state=RANDOM_STATE, perplexity=30, n_iter=1000)
X_tsne_2d = tsne.fit_transform(X_tsne)

plt.figure(figsize=(10, 8))
plt.scatter(X_tsne_2d[y_tsne == 0, 0], X_tsne_2d[y_tsne == 0, 1], label='Genuine (0)', color='#1f77b4', alpha=0.6, edgecolors='w', s=40)
plt.scatter(X_tsne_2d[y_tsne == 1, 0], X_tsne_2d[y_tsne == 1, 1], label='Fraud (1)', color='#d62728', alpha=0.9, edgecolors='k', s=60)
plt.title('t-SNE 2D Spatial Layout of Transactions')
plt.xlabel('t-SNE Dimension 1')
plt.ylabel('t-SNE Dimension 2')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
# Scale the 'Time' and 'Amount' columns
scaler = StandardScaler()
df['Scaled_Amount'] = scaler.fit_transform(df['Amount'].values.reshape(-1, 1))
df['Scaled_Time'] = scaler.fit_transform(df['Time'].values.reshape(-1, 1))

# Drop original Time and Amount columns as we now have scaled versions
df.drop(['Time', 'Amount'], axis=1, inplace=True)

# Define Features (X) and Target (y)
X = df.drop('Class', axis=1)
y = df['Class']

# Split the dataset into train and test sets (Stratified to maintain class balance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
print(f"Train label distribution:\n{y_train.value_counts()}")
````

````python
# Train a baseline Logistic Regression model on the imbalanced training data
baseline_model = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
baseline_model.fit(X_train, y_train)

# Evaluate the baseline model on the test set
y_pred_baseline = baseline_model.predict(X_test)

print("=== Baseline Logistic Regression (Imbalanced Data) ===")
print(classification_report(y_test, y_pred_baseline))
print(f"Area Under PR Curve: {average_precision_score(y_test, y_pred_baseline):.4f}")
````

````python
# Define an imbalanced-learn Pipeline to perform SMOTE and train Logistic Regression
pipeline = Pipeline([
    ('smote', SMOTE(random_state=RANDOM_STATE)),
    ('classifier', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
])

# Fit the pipeline on the training data
pipeline.fit(X_train, y_train)

# Predict on the test set
y_pred_smote = pipeline.predict(X_test)

print("=== Logistic Regression with SMOTE (Balanced Data) ===")
print(classification_report(y_test, y_pred_smote))
print(f"Area Under PR Curve: {average_precision_score(y_test, y_pred_smote):.4f}")
````

````python
from sklearn.ensemble import RandomForestClassifier

# Initialize and train the Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1)
rf_model.fit(X_train, y_train)

# Predict on the test set
y_pred_rf = rf_model.predict(X_test)

print("=== Random Forest Classifier ===")
print(classification_report(y_test, y_pred_rf))
print(f"Area Under PR Curve: {average_precision_score(y_test, y_pred_rf):.4f}")
````

## Prepare Files for Streamlit and GitHub Deployment
To deploy the web app, we first need to save our trained Random Forest model and scaling configurations, then write out the Streamlit user interface file (`app.py`).

````python
# Save model and scaler objects
joblib.dump(rf_model, 'fraud_rf_model.joblib')
joblib.dump(scaler, 'scaler.joblib')
print("Model and Scaler successfully saved!")
````

````python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib

st.set_page_config(page_title="Credit Card Fraud Detector", layout="wide")

st.title("💳 Credit Card Fraud Detection Portal")
st.write("Enter the transaction details below to verify if the transaction is legitimate or likely fraudulent.")

# Load artifacts safely
@st.cache_resource
def load_artifacts():
    model = joblib.load('fraud_rf_model.joblib')
    scaler = joblib.load('scaler.joblib')
    return model, scaler

try:
    model, scaler = load_artifacts()
except Exception as e:
    st.error(f"Error loading models. Please ensure 'fraud_rf_model.joblib' and 'scaler.joblib' are in the same folder. Error: {e}")
    st.stop()

# Split input configuration into columns
col1, col2, col3 = st.columns(3)

with col1:
    raw_time = st.number_input("Transaction Time (Seconds)", min_value=0.0, value=100.0, step=1.0)
    raw_amount = st.number_input("Transaction Amount ($)", min_value=0.0, value=50.0, step=0.01)

with col2:
    st.write("**Principal Component Anonymized Features (V1-V14)**")
    v_inputs = {}
    for i in range(1, 15):
        v_inputs[f'V{i}'] = st.number_input(f'V{i}', value=0.0, step=0.1)

with col3:
    st.write("**Principal Component Anonymized Features (V15-V28)**")
    for i in range(15, 29):
        v_inputs[f'V{i}'] = st.number_input(f'V{i}', value=0.0, step=0.1)

if st.button("Run Fraud Audit Prediction", type="primary"):
    # Correctly scale the Amount and Time variables
    scaled_amount = scaler.transform(np.array([[raw_amount]]))[0][0]
    # We fit scaled time manually based on how scaler was used during development
    # Note: If time was scaled separately, we use the same scaler or transformation logic:
    scaled_time = scaler.transform(np.array([[raw_time]]))[0][0] # Adjust according to dataset constraints

    # Reconstruct exact feature order expected by our random forest classifier
    feature_order = [f'V{i}' for i in range(1, 29)] + ['Scaled_Amount', 'Scaled_Time']

    input_dict = {f'V{i}': [v_inputs[f'V{i}']] for i in range(1, 29)}
    input_dict['Scaled_Amount'] = [scaled_amount]
    input_dict['Scaled_Time'] = [scaled_time]

    input_df = pd.DataFrame(input_dict)[feature_order]

    # Classify
    prediction = model.predict(input_df)[0]
    probability = model.predict_proba(input_df)[0][1]

    st.subheader("Audit Verdict:")
    if prediction == 1:
        st.error(f"🚨 ALERT: FRAUD DETECTED (Confidence: {probability * 100:.2f}%)")
    else:
        st.success(f"✅ Transaction Validated (Genuine Probability: {(1 - probability) * 100:.2f}%)")
````
